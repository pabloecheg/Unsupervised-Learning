```python
from google.colab import drive # conectar con datos de drive
drive.mount('/content/gdrive')
from ntscraper import Nitter
import pandas as pd
```

Con Nitter se pueden extraer Tweets de hashtag, usuarios o terminos

```python
scraper= Nitter(log_level = 1, skip_instance_check= False) # establecer conexión
```

```python
help(scraper.get_tweets)
```

```python
tweets= scraper.get_tweets("PetroElEscamoso", mode="hashtag", number=300)
```

```python
data = {
    'link':[],
    'text':[],
    'user':[],
    'likes':[],
    'quotes':[],
    'retweets':[],
    'comments':[]
}

for tweet in tweets['tweets']:
    data['link'].append(tweet['link'])
    data['text'].append(tweet['text'])
    data['user'].append(tweet['user']['name'])
    data['likes'].append(tweet['stats']['likes'])
    data['quotes'].append(tweet['stats']['quotes'])
    data['retweets'].append(tweet['stats']['retweets'])
    data['comments'].append(tweet['stats']['comments'])
```

```python

pd.DataFrame(data)
```
